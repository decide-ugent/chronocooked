```python
# CNN + LSTM - train for time production task 
target_duration = 9
ent_coeff = 0
timestep = 200
gamma = 0.99
oven_delay_prob = 0.0
overcooked_reward=0.5 # default
undercooked_reward = 0.1  # default 
target_kl = 0.03
tkl=0.03 # for model save path 


env = TimeProductionUncertainity(oven_duration=target_duration, max_timesteps=timestep, oven_delay_prob= oven_delay_prob)

policy_kwargs = dict(
    features_extractor_class=CustomCNN,
    features_extractor_kwargs=dict(features_dim=64, n_channels=6),
    shared_lstm=True,  # share LSTM between actor & critic
    enable_critic_lstm = False,
    lstm_hidden_size=125,
    n_lstm_layers=1,
    net_arch = dict(pi=[64], vf=[64])
)

model_save_path = f"models/cl_uncertainity_tp_T{target_duration}_runder{undercooked_reward}_rover{overcooked_reward}_cnnout64_ts{timestep}_entcoef{ent_coeff}_nepoch20_kl{tkl}_sharedlstm125_mlp64_gamma{gamma}_logs/"

model = RecurrentPPO("MlpLstmPolicy", env, policy_kwargs=policy_kwargs, verbose=1, n_steps=1000, n_epochs=20, target_kl=target_kl, ent_coef=ent_coeff,
                     gamma=gamma,
                     tensorboard_log=model_save_path)

checkpoint_callback = CheckpointCallback(
        save_freq=50000,
        save_path=model_save_path,
        name_prefix="rl_model",
        save_replay_buffer=True,
        save_vecnormalize=True,
    )
```

```python
# Train model
total_timesteps = 100000
model.learn(total_timesteps=total_timesteps, callback=checkpoint_callback,  tb_log_name="first_run", reset_num_timesteps=False)
```

```python
# test time policy
env_test = TimeProductionUncertainity(oven_duration=target_duration, max_timesteps=timestep, oven_delay_prob= oven_delay_prob)
# env_test = TimeProductionUncertainity(oven_duration=target_duration, max_timesteps=timestep, oven_delay_prob= 0.5)
model = RecurrentPPO.load(f"{model_save_path}/rl_model_200000_steps.zip", env=env_test)
# model = RecurrentPPO.load("models/overcooked_wait_singleagent_T8_v0_logs/rl_model_conttrain_98000_steps.zip", env=env_test)
# print(f)
# delayed tp 
# print(f)
seed = 21
obs, _ = env_test.reset(seed=seed)
visual_obs = np.sum(obs,axis=2)
obs_concat = [visual_obs]
act_concat = []
reward_concat = []
# print("oven timers", env_test.oven1_timer, env_test.oven2_timer)
rewards=0
states = None
for _ in range(timestep + 2):
    # obs = obs + 300
    action, states = model.predict(obs, state=states, deterministic=True) 
    act_concat.append(action)
    # print(visual_obs)
    # print("action",action)
    obs, reward, done, _ ,_= env_test.step(action)
    visual_obs = np.sum(obs,axis=2)
    obs_concat.append(visual_obs)

    # print("reward",reward)
    # print("oven timer", env_test.oven_timer)
    
    rewards += reward
    reward_concat.append(reward)
    if done:
        obs,_ = env_test.reset()

        print("rewards",rewards)
        break
arr = np.array(obs_concat)  # (200, 5, 3)

n_rows = 1
n_cols = len(arr) // n_rows -1  # 40

fig, axes = plt.subplots(n_rows, n_cols, figsize=(15,2))

for i, ax in enumerate(axes.flat):
    ax.imshow(arr[i], cmap = 'turbo', vmin=0, vmax=10, interpolation='nearest', aspect='auto')
    # ax.text(0.5, -0.6, str(act_concat[i]), fontsize=12)
    # ax.text(0.2, 5.6, str(reward_concat[i]), fontsize=12)
    if reward_concat[i] != 0:
        ax.set_title(f"A:{act_concat[i]} R:{reward_concat[i]}", fontsize=9, fontdict={"color":"red"})
    else:
        ax.set_title(f"A:{act_concat[i]}", fontsize=9)
    # x = ax.set_xticks(x_ticks, [f"{int(l[0])} , {str(l[1])}, {str(l[2])} " for l in zip(act_concat, range(len(act_concat)), reward_concat)], rotation =90, ha='right', fontsize=10)

    ax.axis('off')  # hide axes for clarity

plt.tight_layout()
plt.show()
```
