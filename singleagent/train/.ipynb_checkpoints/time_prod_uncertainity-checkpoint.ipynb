```python
import sys

import numpy as np

sys.path.append("/project_ghent/chronocooked")

import os
from sb3_contrib import RecurrentPPO

from stable_baselines3.common.callbacks import CheckpointCallback, BaseCallback, CallbackList

# feature extractors
from singleagent.sb3_utils.feature_extractors.custom_cnn_simplified import CustomCNN

# load env
from singleagent.rl_environments.time_production_uncertainity import TimeProductionUncertainity


class CurriculumCallback(BaseCallback):
    def __init__(self, thresholds, env_delay, checkpoint_callback, window_size=100, verbose=1):
        super().__init__(verbose)
        self.thresholds = thresholds
        self.env_delay = env_delay
        self.window_size = window_size
        self.rewards = []
        self.current_level = 0
        self.checkpoint_callback = checkpoint_callback
        self.prev_old_updates = 0


    def _on_step(self) -> bool:
        return True

    def _on_rollout_end(self) -> None:
        ep_info_buffer = self.model.ep_info_buffer
        n_updates = self.logger.name_to_value.get('train/n_updates')
        # print("old n updates",self.prev_old_updates, n_updates , self.prev_old_updates )
        # print("save freq", self.checkpoint_callback.save_freq)
        # print(len(ep_info_buffer))

        if len(ep_info_buffer) >= self.window_size:
            avg_reward = np.mean([ep_info["r"] for ep_info in ep_info_buffer])
        else:
            return

        # if n_updates is None:
        #     return

        # switch to next env
        if (self.current_level < len(self.thresholds)) and (avg_reward >= self.thresholds[self.current_level] ):

            self.current_level += 1
            self.rewards = []

            self.model.policy.optimizer.state.clear()
            self.model.ep_info_buffer.clear()
            self.model.set_env(self.env_delay)
            obs = self.model.env.reset()
            self.model._last_obs = obs
   

            self.checkpoint_callback.name_prefix = f"rl_model_env{self.env_delay.oven_delay_prob}"
            self.checkpoint_callback.save_freq = 20000
            self.prev_old_updates = n_updates

            print(f"Switched to level {self.current_level}")


target_duration = 9
oven_delay_prob = 1
ent_coeff = 0.05

timestep = 200
g=0.99
overcooked_reward=0.5 # default
undercooked_reward = 0.1  # default

for log in [1]:
    for lstm_size in [8,125,256]:

        models_folder = "/project_ghent/chronocooked/singleagent/train/tp_uncertainity_cl_modelsv4"

        env = TimeProductionUncertainity(oven_duration=target_duration, max_timesteps=timestep,
                                         oven_delay_prob= 0, overcooked_reward=overcooked_reward,
                                         undercooked_reward=undercooked_reward)
        
        env_delay = TimeProductionUncertainity(oven_duration=target_duration, max_timesteps=timestep,
                                         oven_delay_prob= oven_delay_prob, overcooked_reward=overcooked_reward,
                                         undercooked_reward=undercooked_reward)

        thresholds = [0.98]

        policy_kwargs = dict(
            features_extractor_class=CustomCNN,
            features_extractor_kwargs=dict(features_dim=64, n_channels=6),
            shared_lstm=True,  # share LSTM between actor & critic
            enable_critic_lstm = False,
            lstm_hidden_size=lstm_size,
            n_lstm_layers=1,
            net_arch = dict(pi=[64], vf=[64])
        )

        model_save_path = f"models/v4_cl_uncertainty_tp_T{target_duration}_runder{undercooked_reward}_rover{overcooked_reward}_delay{oven_delay_prob}_cnnout64_ts{timestep}_entcoef{ent_coeff}_nepoch20_sharedlstm{lstm_size}_mlp64_gamma{g}_logs{log}/"

        checkpoint_callback = CheckpointCallback(
                save_freq=50000,
                save_path=model_save_path,
                name_prefix="rl_model_env0",
                save_replay_buffer=True,
                save_vecnormalize=True,
            )

        callback = CallbackList([
            checkpoint_callback,
            CurriculumCallback(thresholds=thresholds,env_delay = env_delay, checkpoint_callback = checkpoint_callback)
        ])

        model = RecurrentPPO("MlpLstmPolicy", env, policy_kwargs=policy_kwargs, verbose=1, n_steps=1000, n_epochs=20, ent_coef=ent_coeff,
                             gamma = g, tensorboard_log=model_save_path)
        # Train model
        total_timesteps = 960000
        model.learn(total_timesteps=total_timesteps, callback=callback,  tb_log_name="first_run", reset_num_timesteps=False)
```

```python
# task performance
# seeds
# map seed to each starting point
env = TimeProductionUncertainity(max_timesteps=200)
seeds = {(int(ax),int(ay)): -1 for ax, ay in zip(np.where(env.reset(seed=0)[0][:,:,0] ==0)[0],np.where(env.reset(seed=0)[0][:,:,0] ==0)[1])}
for s in range(1,30,1):
    ax, ay = np.where(env.reset(seed=s)[0][:,:,1] ==1)
    if seeds[int(ax[0]),int(ay[0])] == -1:
        seeds[int(ax[0]),int(ay[0])] = s

def get_eprew_across_training(tb_event_file,max_training_ts_env0 ):
    ea = event_accumulator.EventAccumulator(tb_event_file)
    ea.Reload()
    eprew = ea.Scalars("rollout/ep_rew_mean") 
    eprew_dict = {a_kl.step:a_kl.value for a_kl in eprew if (a_kl.step % 20000 == 0) & (a_kl.step>max_training_ts_env0)}
    return eprew_dict

def get_eplen_across_training(tb_event_file, max_training_ts_env0):
    ea = event_accumulator.EventAccumulator(tb_event_file)
    ea.Reload()
    eplen = ea.Scalars("rollout/ep_len_mean") 
    eplen_dict = {a_kl.step:a_kl.value for a_kl in eplen if (a_kl.step % 20000 == 0) & (a_kl.step>max_training_ts_env0)}
    return eplen_dict

traj_labels = {"adj_oven_interact": 1, "adj_oven_other":2, 
               "adj_button_interact": 3, "adj_button_other":4,
              "adj_other_interact":5, "other":0}


def get_traj_label(adjacent_cells, action, env):
    if env.oven_pos in adjacent_cells.keys(): 
        if action==5:
            traj_label=1
        else:
            traj_label = 2
    elif env.manual_ovenstop_pos in adjacent_cells.keys():
        if action==5:
            traj_label=3
        else:
            traj_label = 4  

    elif action == 5:
        traj_label=5
    else:
        traj_label=0
    return traj_label
```

```python
target_durations = [9]
undercooked_reward = 0.1
overcooked_reward = 0.5
oven_delay_probabilities = [1,0.2]
timestep = 200
ent_coeff = 0.05
g=0.99

lstm_sizes=[8]
delay_rew_threshold = 0.98
models_folder = "/project_ghent/chronocooked/singleagent/train/tp_uncertainity_cl_modelsv4"
models_folder = "models"
rows=[]
for target_duration in target_durations:
    for oven_delay_prob in oven_delay_probabilities:
        for lstm_size in lstm_sizes:
            for log in [1]:
                if lstm_size == None:
                    model_file = f"v4_cl_uncertainty_tp_T{target_duration}_runder{undercooked_reward}_rover{overcooked_reward}_delay{oven_delay_prob}_cnnout64_ts{timestep}_entcoef{ent_coeff}_nepoch20_mlp64_gamma{g}_logs{log}/"
                    label = "CNN"
                else:
                    model_file = f"v4_cl_uncertainty_tp_T{target_duration}_runder{undercooked_reward}_rover{overcooked_reward}_delay{oven_delay_prob}_cnnout64_ts{timestep}_entcoef{ent_coeff}_nepoch20_sharedlstm{lstm_size}_mlp64_gamma{g}_logs{log}/"
                    label = f"LSTM{lstm_size}"

                model_files = os.listdir(f"{models_folder}/{model_file}")
                max_training_ts_env0 = max([int(model_file.split("env")[1].split("_")[1])  if ("env" in model_file) and (model_file.split("env")[1].split("_")[0] == "0") else 0 for model_file in model_files])
                max_timesteps = max([int(model_file.split("env")[1].split("_")[1])  if ("env" in model_file) and (model_file.split("env")[1].split("_")[0] == "0.2") else 0 for model_file in model_files])
                if max_training_ts_env0 <= max_timesteps:
                    print("delay phase not reached", target_duration, oven_delay_prob, lstm_size, log)
                    continue 

                # get first successful training ts in delay phase
                tb_event_file = f"{models_folder}/{model_file}/first_run_0"
                # eprew_dict = get_eprew_across_training(tb_event_file, max_training_ts_env0)
                eplen_dict = get_eplen_across_training(tb_event_file, max_training_ts_env0)
                # min_success_delay_ts = min([ts for ts, rew in eprew_dict.items() if rew>delay_rew_threshold])
                min_success_delay_ts = min([ts if eplen<40 else 1000000 for ts, eplen in eplen_dict.items() ])
                # print(min_success_delay_ts)
                # for ts in range(max_training_ts_env0, min_success_delay_ts, 40000):
                for ts in range(max_training_ts_env0, 960000, 20000):
                    
                    oven_delay_prob2 = 1
                    env_test = TimeProductionUncertainity(oven_duration=target_duration, max_timesteps=timestep, oven_delay_prob= oven_delay_prob2)
                    try:
                        if lstm_size == None:
                            model= PPO.load(f"{models_folder}/{model_file}/rl_model_env{oven_delay_prob}_{ts}_steps.zip")
                        else:
                            model= RecurrentPPO.load(f"{models_folder}/{model_file}/rl_model_env{oven_delay_prob}_{ts}_steps.zip")
                        
                    except:
                        print("could not load",f"{models_folder}/{model_file}/rl_model_env{oven_delay_prob}_{ts}_steps.zip")
                        continue
                        
                    for s_i, seed in enumerate(seeds.values()):               
                        model.set_env(env_test)
                        obs, _ = env_test.reset(seed=seed)
                        visual_obs = np.sum(obs,axis=2)
                        obs_concat = [visual_obs]
                        act_concat = []
                        reward_concat = []
                        oven_timers = []
                        timing_act_labels = []
                        # print("oven timers", env_test.oven1_timer, env_test.oven2_timer)
                        rewards=0
                        states = None
                        
                        for i in range(timestep + 2):
                            adjacent_cells = env_test.get_adjacent_cells(env_test.agent_position)
                            
                            # obs = obs + 300
                            action, states = model.predict(obs, state=states, deterministic=True) 
                            act_concat.append(action)
                            oven_timers.append(env_test.oven_timer)                
                            
                            # print(visual_obs)
                            # print("action",action)
                            obs, reward, done, _ ,_= env_test.step(action)
                            visual_obs = np.sum(obs,axis=2)
                            obs_concat.append(visual_obs)

                            if env_test.oven_timer > -1:

                                timing_act_labels.append(get_traj_label(adjacent_cells, action, env))
                        
                            # print("reward",reward)
                            # print("oven timer", env_test.oven_timer)
                            
                            rewards += reward
                            reward_concat.append(reward)
                            if done:
                                reward_concat.append(reward)
                                act_concat.append(0)
                                oven_timers.append(env_test.oven_timer) 
                                if (reward == 1 and oven_delay_prob2 == 0) or (reward == 0.5 and oven_delay_prob2 == 1):
                                    success = 1
                                else:
                                    success = 0
                                # obs,_ = env_test.reset()
                                rows.append({"TD":target_duration, "log": log, "ts":ts, "seed":s_i, "model":label,"oven_delay_prob": oven_delay_prob, "oven_delay_prob2": oven_delay_prob2, "manual_ovenstop_time": env_test.manual_ovenstop_time,
                                                    "oven_off_time": env_test.oven_off_time  ,"rewards":rewards, "total_timesteps": i+1,
                                                    "success": success, "timing_phase_actions": [b for a, b in zip(oven_timers, act_concat) if a != -1],
                                             "timing_act_labels": timing_act_labels
                                            })
                                break
            break
        break
    break

task_performance_df = pd.DataFrame(rows)     
```

```python
import sys

import numpy as np

sys.path.append("/project_ghent/chronocooked")

import os
from sb3_contrib import RecurrentPPO

from stable_baselines3.common.callbacks import CheckpointCallback, BaseCallback, CallbackList

# feature extractors
from singleagent.sb3_utils.feature_extractors.custom_cnn_simplified import CustomCNN

# load env
from singleagent.rl_environments.time_production_uncertainity import TimeProductionUncertainity

target_duration = 4
oven_delay_prob = 0.2
ent_coeff = 0.05

timestep = 200
g=0.99
log=1
overcooked_reward=0.5 # default
undercooked_reward = 0.1  # default
lstm_size = 125

class CurriculumCallback(BaseCallback):
    def __init__(self, thresholds, oven_delay_prob, checkpoint_callback, window_size=100, verbose=1):
        super().__init__(verbose)
        self.thresholds = thresholds
        self.oven_delay_prob = oven_delay_prob
        self.window_size = window_size
        self.rewards = []
        self.current_level = 0
        self.checkpoint_callback = checkpoint_callback
        self.prev_old_updates = 0
        self.n_update_threshold = 1000


    def _on_step(self) -> bool:
        return True

    def _on_rollout_end(self) -> None:
        ep_info_buffer = self.model.ep_info_buffer
        n_updates = self.logger.name_to_value.get('train/n_updates')
        # print("old n updates",self.prev_old_updates, n_updates , self.prev_old_updates )
        # print("save freq", self.checkpoint_callback.save_freq)
        # print(len(ep_info_buffer))

        if len(ep_info_buffer) >= self.window_size:
            avg_reward = np.mean([ep_info["r"] for ep_info in ep_info_buffer])
        else:
            return

        # if n_updates is None:
        #     return

        # switch to next env
        if (self.current_level < len(self.thresholds)) and (avg_reward >= self.thresholds[self.current_level] ) and ((n_updates - self.prev_old_updates) >= self.n_update_threshold):

            self.current_level += 1
            self.rewards = []

            for env in self.training_env.envs:
                env.oven_delay_prob = self.oven_delay_prob

                print("new oven delay prob",env.oven_delay_prob)

            self.checkpoint_callback.name_prefix = f"rl_model_env{self.oven_delay_prob}_{self.thresholds[self.current_level-1]}"
            self.checkpoint_callback.save_freq = 20000
            self.prev_old_updates = n_updates

            print(f"Switched to level {self.current_level}")




# for log in [1,2,3]:
#     for lstm_size in [8,125,256]:

env = TimeProductionUncertainity(oven_duration=target_duration, max_timesteps=timestep,
                                 oven_delay_prob= 0, overcooked_reward=overcooked_reward,
                                 undercooked_reward=undercooked_reward)

thresholds = [0.97,0.98,1]

policy_kwargs = dict(
    features_extractor_class=CustomCNN,
    features_extractor_kwargs=dict(features_dim=64, n_channels=6),
    shared_lstm=True,  # share LSTM between actor & critic
    enable_critic_lstm = False,
    lstm_hidden_size=lstm_size,
    n_lstm_layers=1,
    net_arch = dict(pi=[64], vf=[64])
)

model_save_path = f"models/v3_cl_uncertainity_tp_T{target_duration}_runder{undercooked_reward}_rover{overcooked_reward}_delay{oven_delay_prob}_cnnout64_ts{timestep}_entcoef{ent_coeff}_nepoch20_sharedlstm{lstm_size}_mlp64_gamma{g}_logs2/"

checkpoint_callback = CheckpointCallback(
        save_freq=50000,
        save_path=model_save_path,
        name_prefix="rl_model_env0",
        save_replay_buffer=True,
        save_vecnormalize=True,
    )

callback = CallbackList([
    checkpoint_callback,
    CurriculumCallback(thresholds=thresholds,oven_delay_prob = oven_delay_prob, checkpoint_callback = checkpoint_callback)
])

# model = RecurrentPPO("MlpLstmPolicy", env, policy_kwargs=policy_kwargs, verbose=1, n_steps=1000, n_epochs=20, ent_coef=ent_coeff,
#                      gamma = g, tensorboard_log=model_save_path)
# Train model

model = RecurrentPPO.load(f"{model_save_path}/rl_model_env0_{100000}_steps.zip", env)
total_timesteps = 960000
model.learn(total_timesteps=total_timesteps, callback=callback,  tb_log_name="first_run", reset_num_timesteps=False)
```
