```python
import torch
import torch.nn as nn
from spikingjelly.activation_based import neuron, layer, functional, surrogate
from stable_baselines3.common.torch_layers import BaseFeaturesExtractor


T = 24  # SNN timesteps (rate coding window)

class SNNExtractor(BaseFeaturesExtractor):
    def __init__(self, observation_space, features_dim=32):
        super().__init__(observation_space, features_dim)

        # obs_flatten_dim = observation_space.shape[0] * observation_space.shape[1] * observation_space.shape[2]

        self.flatten = nn.Flatten()
        # self.linear = nn.Sequential(nn.Linear(obs_flatten_dim, features_dim))

        with torch.no_grad():
            sample = torch.as_tensor(observation_space.sample()[None]).float()
            obs_flatten_dim = sample.view(1, -1).shape[1]

        self.net = nn.Sequential(
            layer.Linear(obs_flatten_dim, features_dim),
            neuron.LIFNode(surrogate_function=surrogate.ATan()),
            layer.Linear(features_dim, features_dim),
            neuron.LIFNode(surrogate_function=surrogate.ATan()),
        )
        functional.set_step_mode(self.net, step_mode='m')  # multi-step

    def forward(self, obs):
        obs = self.flatten(obs)
        obs = obs / 20
        functional.reset_net(self.net)
        x = (torch.rand(T, *obs.shape) < obs.unsqueeze(0)).float()
        spikes = self.net(x)        # (T, batch, features_dim)
        return spikes.mean(0)  
```
