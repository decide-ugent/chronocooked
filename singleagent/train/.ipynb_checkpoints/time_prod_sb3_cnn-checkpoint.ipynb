```python
target_duration = 24
```

```python
if __name__ == "__main__":

    env = ChronoCookedTimeProduction(oven_duration=target_duration)
    obs, _ = env.reset(seed=4)

    visual_obs = np.sum(obs,axis=2)

    # plt.imshow(visual_obs, cmap='viridis')
    # plt.show()
    print("oven timers", env.oven1_timer)
    rewards = 0
    while True:
        action = int(input("Enter action: "))
        obs, reward, done, _,_= env.step(action)
        visual_obs = np.sum(obs,axis=2)
        # plt.imshow(visual_obs)  # Update image data
        # plt.show()
        print("oven_timers",env.oven1_timer)
        print("reward",reward, "step", env.timestep)
        print("points", env.num_soups_delivered)
        print("adjacent cell", env.get_adjacent_cells(env.agent_position))
        print(env.oven1_pos, env.grid_map[env.agent_position[0],env.agent_position[1],2], env.oven1_timer )
        print(visual_obs)
        rewards += reward
        if done:
            print(rewards,"reward")
            break
```

```python

env = ChronoCookedTimeProduction(oven_duration=target_duration, max_timesteps=200)

policy_kwargs = dict(
    features_extractor_class=CustomCNN,
    features_extractor_kwargs=dict(features_dim=64, n_channels=6),
    net_arch = dict(pi=[64], vf=[64])
)

model = PPO("MlpPolicy", env, policy_kwargs=policy_kwargs, verbose=1, n_steps=1000, n_epochs=20, target_kl=0.03, 
                     tensorboard_log=f"models/singletask_onetrial_T{target_duration}_cnnoutput64_timestep200_entcoef00_nepoch20_kl03_mlp64_logs/")
```

```python
model.policy
```

```python
# n_step = 1000, means n_ep * len_of_ep = 1000. So before an update, n_ep are collected. 
# n_epoch = 20 means updates are done 20 times using the 1000 steps.
# batch_size = n_steps * n_env. so also 1000. 

# small batch_size = n_steps = 128 - does it make trainig unstable?
# larger batch_size = n_steps = 1000 - more stable ?
```

```python
checkpoint_callback = CheckpointCallback(
        save_freq=5000,
        save_path=f"models/singletask_onetrial_T{target_duration}_cnnoutput64_timestep200_entcoef00_nepoch20_kl03_mlp64_logs/",
        name_prefix="rl_model",
        save_replay_buffer=True,
        save_vecnormalize=True,
    )
```

```python
# Train model
total_timesteps = 500000
model.learn(total_timesteps=total_timesteps, callback=checkpoint_callback,  tb_log_name="first_run", reset_num_timesteps=False)
```
